```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#zadatak 1

np.random.seed(42)                
N = 10000

alfa = 2.35
M_min, M_max = 0.08, 200.0         

u = np.random.random(N)            
a = 1 - alfa
masa = (M_min**a + u*(M_max**a - M_min**a))**(1/a)

print('Najmanja masa:', masa.min(), 'Msun')
print('Najveca masa:', masa.max(), 'Msun')
print('Medijana:', np.median(masa), 'Msun')

bins = np.logspace(np.log10(M_min), np.log10(masa.max()), 40)
plt.figure(figsize=(8, 5))
plt.hist(masa, bins=bins, edgecolor='black')
plt.xscale('log'); plt.yscale('log')
plt.xlabel('Masa zvezda')
plt.ylabel('Broj zvezda')
plt.title('Semplovane mase (Salpeterova IMF)')
plt.show()

print (" \n Mase su semplovane inverse transform metodom: CDF Salpeterove IMF je invertovana analiticki i primenjena na uniformne slucajne brojeve. \n " \
"Raspodela mase je jako nagnuta ka malim masama: medijana je oko 0.13 Msun, a oko 92% zvezda ima M < 0.5 Msun. \n " \
"Najveca semplovana masa je ~32 Msun, a ne 200 Msun, jer su zvezde velikih masa toliko retke da u uzorku od 10000 zvezda nijedna ne dostigne gornju granicu. \n " \
"Histogram u log-log skali prati pravu sa nagibom -2.35, sto potvrdjuje da je sempl dobar.")
```

```python
# zadatak 2

L = masa**3.5                              

M_sunce = 4.74                             
M_abs = M_sunce - 2.5*np.log10(L)          

print('L: od', L.min(), 'do', L.max(), 'Lsun')
print('M_abs: od', M_abs.min(), 'do', M_abs.max())
print('Srednja M_abs: ', M_abs.mean())

print(" \n Luminoznost je odredjena relacijom masa-luminoznost L ~ M^3.5, a apsolutna magnituda formulom M = M_sunce - 2.5 log10(L/Lsun). \n" \
" Luminoznosti pokrivaju oko 9 redova velicine (od 1e-4 do 2e5 Lsun), a apsolutne magnitude se krecu od oko -8.4 do oko 14.3. \n " \
"Posto su zvezde malih masa daleko najbrojnije, srednja apsolutna magnituda je velika (11.6), odnosno tipicna zvezda je slabo luminozna patuljak. \n " \
"Masivne zvezde su retke, ali ekstremno sjajne.")
```

```python
# zadatak 3

d_max = 10000.0                            
d = d_max*np.random.random(N)**(1/3)       

m = M_abs + 5*np.log10(d/10)               

plt.figure(figsize=(8, 5))
plt.hist(m, bins=60, edgecolor='black')
plt.xlabel('Prividna magnituda m')
plt.ylabel('Broj zvezda')
plt.title('Histogram prividnih magnituda')
plt.show()

print(" Udaljenosti su semplovane kao d = d_max * u^(1/3), jer je za uniformnu raspodelu u sferi P(<d) = (d/d_max)^3. \n " \
"Zbog toga broj zvezda raste sa udaljenoscu, a najbliza zvezda je tek na ~0.5 kpc (u prvih 1 kpc ima oko 10 zvezda). \n " \
"Prividne magnitude se krecu od ~6 do ~29, a medijana je ~26.6. \n" \
" Dakle vecina zvezda je veoma slaba, jer su i slabe po sebi (velika M_abs) i daleke.")
```

```python
# zadatak 4

m_lim_lista = [15, 18, 22]
boje = ['red', 'green', 'blue']

fig, axs = plt.subplots(1, 2, figsize=(13, 5))
bins_d = np.linspace(0, d_max, 21)

for m_lim, boja in zip(m_lim_lista, boje):
    det = m < m_lim
    print('m_lim = {}: detektovano {} od {} zvezda ({:.1f} %)'.format(
          m_lim, det.sum(), N, 100*det.mean()))

    axs[0].hist(d[det]/1000, bins=bins_d/1000, histtype='step', color=boja,
                linewidth=2, label=f'm < {m_lim} ({det.sum()} zvezda)')

    d_sort = np.sort(d[det])
    axs[1].plot(d_sort/1000, np.arange(1, len(d_sort)+1), color=boja,
                linewidth=2, label=f'm < {m_lim}')

d_sort_svi = np.sort(d)
axs[1].plot(d_sort_svi/1000, np.arange(1, N+1), color='gray', linestyle='--', label='sve zvezde')

axs[0].set_xlabel('Udaljenost d [kpc]')
axs[0].set_ylabel('Broj detektovanih zvezda')
axs[0].set_title('Raspodela detektovanih zvezda po udaljenosti')
axs[0].legend()

axs[1].set_xlabel('Udaljenost d [kpc]')
axs[1].set_ylabel('Broj zvezda bliže od d')
axs[1].set_yscale('log')
axs[1].set_title('Kumulativni broj')
axs[1].legend()
axs[1].grid(alpha=0.3)
plt.tight_layout()
plt.show()

print(" \n Sa povecanjem granicne magnitude detektuje se vise zvezda i to u celom opsegu udaljenosti. \n " \
"Ukupan broj zvezda raste kao d^3, a broj detektovanih sporije, jer sa udaljenoscu slabije zvezde ispadaju iz uzorka. \n " \
"Za m_lim = 15 i 18 u prvom kiloparseku nema ni jedne detektovane zvezde jer tamo gotovo da nema zvezda, a detektovane su pretezno one daleko.")
```

```python
# zadatak 5

fig, axs = plt.subplots(1, 3, figsize=(16, 5), sharey=True)
d_linija = np.logspace(1, np.log10(d_max), 300)

for ax, m_lim in zip(axs, m_lim_lista):
    det = m < m_lim
    ax.scatter(d[~det]/1000, M_abs[~det], s=3, marker='X', color='red', label='nedetektovane')
    ax.scatter(d[det]/1000, M_abs[det], s=6, marker='o', color='green', label='detektovane')
    ax.plot(d_linija/1000, m_lim - 5*np.log10(d_linija/10), 'k--', label='granica detekcije')

    ax.set_xscale('log')
    ax.invert_yaxis()                      
    ax.set_xlabel('Udaljenost d [kpc]')
    ax.set_title(f'$m_{{lim}}$ = {m_lim}')
    ax.grid(alpha=0.3)
    ax.legend(markerscale=2, fontsize=8)

axs[0].set_ylabel('Apsolutna magnituda $M_{abs}$')
plt.tight_layout()
plt.show()

for m_lim in m_lim_lista:
    det = m < m_lim
    print('m_lim =', m_lim, ' srednje M_abs svih zvezda:', round(M_abs.mean(), 2),
          ' detektovanih:', round(M_abs[det].mean(), 2))
    
print(" \n Isprekidana linija je granica detekcije M = m_lim - 5 log10(d/10): sa rastom udaljenosti ona se pomera ka sjajnijim zvezdama, pa daleko vidimo samo luminozne zvezde. \n " \
"To je Malmkvistova neobjektivnost (selekcioni efekat). \n " \
"Srednja apsolutna magnituda detektovanih zvezda je -1.8 (m_lim=15), 1.3 (m_lim=18) i 5.3 (m_lim=22), a svih zvezda 11.6. \n" \
" Najmanja masa detektovane zvezde je 1.8, 0.65 i 0.18 Msun redom. \n" \
" Uzorak ogranicen po prividnoj magnitudi je pristrasan ka luminoznim zvezdama, pa bi na osnovu njega zakljucili da su zvezde sjajnije nego sto zaista jesu.")
```
